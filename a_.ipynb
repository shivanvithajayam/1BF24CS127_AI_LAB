```python
import heapq
def heuristic(state, goal):
    count = 0
    for i in range(9):
        if state[i] != 0 and state[i] != goal[i]:
            count += 1
    return count
def get_neighbors(state):
    neighbors = []
    zero = state.index(0)
    row = zero // 3
    col = zero % 3
    moves = [
        (-1, 0),
        (1, 0),
        (0, -1),
        (0, 1)
    ]
    for dr, dc in moves:
        nr = row + dr
        nc = col + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            new_zero = nr * 3 + nc
            new_state = list(state)
            new_state[zero], new_state[new_zero] = \
                new_state[new_zero], new_state[zero]
            neighbors.append(tuple(new_state))
    return neighbors
def astar(start, goal):
    pq = []
    g = 0
    h = heuristic(start, goal)
    f = g + h
    heapq.heappush(pq, (f, g, start))
    visited = set()
    states = 0

    while pq:
        pq.sort()
        min_f = pq[0][0]
        tied_items = [item for item in pq if item[0] == min_f]
        if len(tied_items) > 1:
            chosen = tied_items[0]
            pq = [item for item in pq if item[0] != min_f or item == chosen]

        f, g, current = heapq.heappop(pq)
        if current in visited:
            continue
        visited.add(current)
        states += 1

        if current == goal:
            return states, g, True

        for next_state in get_neighbors(current):
            if next_state in visited:
                continue
            new_g = g + 1
            h = heuristic(next_state, goal)
            new_f = new_g + h
            heapq.heappush(pq, (new_f, new_g, next_state))

    return states, -1, False

print("Enter Start State:")
start = []
for i in range(3):
    start.extend(map(int, input().split()))
print("Enter Goal State:")
goal = []
for i in range(3):
    goal.extend(map(int, input().split()))
start = tuple(start)
goal = tuple(goal)
states, cost, found = astar(start, goal)
print("\nNumber of states:", states)
if found:
    print("Cost:", cost)
    print("Found: Yes")
else:
    print("Cost: Not found")
    print("Found: No")
```
