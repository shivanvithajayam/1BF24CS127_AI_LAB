```python
def dfs(startState,finalState):
    stack=[startState]
    visited=set()
    while(len(stack)!=0):
        state=stack.pop()
        if(state==finalState): return "Found"
        if(state in visited): continue
        visited.add(state)
        moves=[(-1,0),(1,0),(0,1),(0,-1)]
        zero=state.index("0")
        r=zero//3
        c=zero%3
        for dr,dc in moves:
            nr=r+dr
            nc=c+dc
            if(0<=nr<3 and 0<=nc<3):
                newzero=nr*3+nc
                board=list(state)
                board[zero],board[newzero]=board[newzero],board[zero]
                newstate=''.join(board)
                if(newstate not in visited):
                   stack.append(newstate)
    return "Not Found"

startState=""
finalState=""
print("Enter the start state (3 lines):")
for i in range(3):
    s=input()
    startState+=s
print("Enter the final state (3 lines):")
for i in range(3):
    s=input()
    finalState+=s

print(dfs(startState,finalState))
```
