```python
def depthLimitedSearch(startState, finalState, limit):
    stack = [(startState, 0)] # (state, depth)
    visited = set()

    while stack:
        state, depth = stack.pop()

        if state == finalState: return "Found"
        if state in visited: continue
        visited.add(state)

        if depth < limit:
            # Generate possible moves (same as in DFS)
            moves = [(-1, 0), (1, 0), (0, 1), (0, -1)]
            zero = state.index("0")
            r = zero // 3
            c = zero % 3

            for dr, dc in moves:
                nr = r + dr
                nc = c + dc
                if 0 <= nr < 3 and 0 <= nc < 3:
                    newzero = nr * 3 + nc
                    board = list(state)
                    board[zero], board[newzero] = board[newzero], board[zero]
                    newState = ''.join(board)

                    if newState not in visited: # Only add if not visited to avoid redundant paths in this DLS iteration
                        stack.append((newState, depth + 1))

    return "Not Found at this depth"

def ids(startState, finalState, max_depth=20): # Set a reasonable max_depth to prevent infinite loops
    for limit in range(max_depth + 1):
        print(f"Searching with depth limit: {limit}")
        result = depthLimitedSearch(startState, finalState, limit)
        if result == "Found":
            return "Found"
    return "Not Found (max depth reached)
```
