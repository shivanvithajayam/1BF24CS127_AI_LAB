```python
rooms = {
    'A': input("Enter status of Room A (Dirty/Clean): ").capitalize(),
    'B': input("Enter status of Room B (Dirty/Clean): ").capitalize()
}
location=input("enter location of vaccum cleaner:").upper()
def check_action(location,status):
  if status=="Dirty":
    return "Suck"
  elif location=="A":
    return "Right"
  else:
    return "Left"

print("\nInitial State:")
print("Room A:", rooms['A'])
print("Room B:", rooms['B'])
print("Vacuum Location:", location)
steps=0
while rooms['A']=="Dirty" or rooms['B']=="Dirty":
  status=rooms[location]
  action=check_action(location,status)
  print("\nCurrent Room:", location)
  print("Current Status:", status)
  print("Action:", action)
  if action=="Suck":
    rooms[location]="Clean"
  elif action=="Right":
    location='B'
  else:
    location='A'
  steps+=1
print("\nGoal Reached!")

print("Final State:")
print("Room A:", rooms['A'])
print("Room B:", rooms['B'])
print("Vacuum Location:", location)

print("Cost=", steps)
```
