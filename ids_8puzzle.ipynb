```python
def dls(state, goal, limit, path, count):
    count[0] += 1
    if state == goal:
        return True
    if limit == 0:
        return False
    zero = state.index(0)
    row, col = divmod(zero, 3)
    moves = [
        (-1, 0),   # Up
        (1, 0),    # Down
        (0, -1),   # Left
        (0, 1)     # Right
    ]
    for dr, dc in moves:
        nr = row + dr
        nc = col + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            new_zero = nr * 3 + nc
            new_state = list(state)
            new_state[zero], new_state[new_zero] = \
                new_state[new_zero], new_state[zero]
            new_state = tuple(new_state)
            if new_state not in path:
                path.append(new_state)

                if dls(new_state, goal, limit - 1, path, count):
                    return True
                path.pop()
    return False
def ids(start, goal, max_depth):
    count = [0]
    for depth in range(max_depth + 1):
        path = [start]
        if dls(start, goal, depth, path, count):
            return count[0], depth, True
    return count[0], -1, False
print("Enter start state (3 lines):")
start = []
for _ in range(3):
    start.extend(map(int, input().split()))
print("Enter goal state (3 lines):")
goal = []
for _ in range(3):
    goal.extend(map(int, input().split()))
limit = int(input("Enter depth limit: "))
start = tuple(start)
goal = tuple(goal)
states, cost, found = ids(start, goal, limit)
print("\nNumber of states:", states)
if found:
    print("Cost:", cost)
    print("Found: Yes")
else:
    print("Cost: Not found")
    print("Found: No")
```
