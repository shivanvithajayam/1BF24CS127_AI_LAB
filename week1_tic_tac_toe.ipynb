```python
board=['']*9
print("TIC TAC TOE")
print("You = X, Computer = O")
print("""
Positions:
1 | 2 | 3
--+---+--
4 | 5 | 6
--+---+--
7 | 8 | 9
""")
def display():
    print()
    print(board[0], "|", board[1], "|", board[2])
    print("--+---+--")
    print(board[3], "|", board[4], "|", board[5])
    print("--+---+--")
    print(board[6], "|", board[7], "|", board[8])
    print()
def check_winner(symbol):

    winning_positions = [(0, 1, 2),(3, 4, 5),(6, 7, 8),(0, 3, 6),(1, 4,                         7),(2, 5, 8),(0, 4, 8),(2, 4, 6)]
    for a, b, c in winning_positions:
        if board[a] == symbol and board[b] == symbol and board[c] ==symbol:

          return True

    return False

def board_full():
  return '' not in board
def human_move():
  while True:
    try:
      position=int(input("Enter position 1-9:"))-1
      if position<0 or position>8:
        print("Enter a number between 1 and 9.")
      elif board[position]!='':
        print("Position already occupied")
      else:
        board[position] = 'X'
        break
    except ValueError:
        print("Please enter a valid number.")
def computer_move():
  for i in range(9):
    if board[i]=='':
      board[i]='O'
      if check_winner('O'):
        return
      board[i]=''
  for i in range(9):
    if board[i]=='':
      board[i]='X'
      if check_winner('X'):


        board[i]='O'
        return

      board[i]=''
  if board[4]=='':
    board[4]='O'
    return
  corners=[0,2,6,8]
  for i in corners:
    if board[i]=='':
      board[i]='O'
      return
  for i in range(9):
    if board[i]=='':
      board[i]='O'

while True:

  human_move()
  display()
  if check_winner('X'):
    display()
    print("Human wins")
    break
  if board_full():
    display()
    print("Game Draw!")
    break
  # Computer turn
  print("Computer's turn...")
  computer_move()
  display()
  if check_winner('O'):

    display()
    print("Computer wins!")
    break
  if board_full():

    display()
```
