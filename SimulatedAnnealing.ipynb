```python
import random
import math

def cost(board):
    conflicts = 0
    for i in range(8):
        for j in range(i + 1, 8):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1
    return conflicts

def simulated_annealing():
    board = list(map(int, input("Enter initial board (0-7): ").split()))
    T = 10.0
    alpha = 0.90

    while T > 0.1:
        if cost(board) == 0:
            break

        next_board = board.copy()
        col = random.randint(0, 7)
        next_row = random.randint(0, 7)

        while next_row == board[col]:
            next_row = random.randint(0, 7)

        next_board[col] = next_row

        delta = cost(next_board) - cost(board)

        if delta < 0:
            board = next_board
        elif random.random() < math.exp(-delta / T):
            board = next_board

        T *= alpha

    return board

board = simulated_annealing()
print("Final Board:", board)
print("Final Cost:", cost(board))
```
