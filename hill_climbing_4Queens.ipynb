```python
class FourQueensProblem:
    def __init__(self, initial_state):
        self.initial_state = initial_state

    def get_value(self, state):
        attacks = 0
        n = len(state)
        for i in range(n):
            for j in range(i + 1, n):
                if state[i] == state[j]:
                    attacks += 1
                elif abs(state[i] - state[j]) == abs(i - j):
                    attacks += 1
        return -attacks

    def get_successors(self, state):
        neighbors = []
        n = len(state)
        for col in range(n):
            for row in range(4):
                if row != state[col]:
                    new_state = list(state)
                    new_state[col] = row
                    neighbors.append(new_state)
        return neighbors

def hill_climbing(problem):
    current_state = problem.initial_state
    print(f"Starting State: {current_state} (Attacks: {-problem.get_value(current_state)})")

    step = 1
    while True:
        neighbors = problem.get_successors(current_state)
        if not neighbors:
            return current_state

        best_neighbor = max(neighbors, key=lambda s: problem.get_value(s))

        if problem.get_value(best_neighbor) <= problem.get_value(current_state):
            print(f"Stuck in Local Maximum/Plateau at step {step}.")
            return current_state

        current_state = best_neighbor
        print(f"Step {step}: Moved to {current_state} (Attacks: {-problem.get_value(current_state)})")
        step += 1

try:
    user_input = input("Enter initial state (4 space-separated integers between 0 and 3, e.g., 2 2 0 3): ")
    initial_state = [int(x) for x in user_input.split()]
    if len(initial_state) != 4 or any(x < 0 or x > 3 for x in initial_state):
        raise ValueError("Please enter exactly 4 integers between 0 and 3.")

    problem = FourQueensProblem(initial_state)
    final_board = hill_climbing(problem)

    print(f"\nFinal Board Result: {final_board}")
    print(f"Final Attack Count: {-problem.get_value(final_board)}")
    if problem.get_value(final_board) == 0:
        print("Success! Found a global optimum.")
    else:
        print("Failed to find a solution (got stuck).")
except ValueError as e:
    print(f"Input Error: {e}")
```
